```python
products = [ "Laptop","Mobile","Headphones","Laptop", "Keyboard", "Mobile"]
print(f"Original list: {products}")
```

```python
categories = ["Electronics","Electronics","Accessories","Electronics","Accessories","Electronics"]
print(f"Categories list: {categories}")
```

```python
categories_set = set(categories)
print(f"Unique categories: {categories_set}")
```

```python
categories_set.add("Gadgets")
print(f"After adding a new category, the unique categories are: {categories_set}")
```

```python
category= input("enter category to search from the set")
is_presnt= category in categories_set
print(f"Is the category '{category}' present in the set? {is_presnt}")
```

```python
length=len(categories_set)
print(f"Number of unique categories: {length}")
```
