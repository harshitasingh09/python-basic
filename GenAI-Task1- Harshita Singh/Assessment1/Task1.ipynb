## List

```python
products= ["mouse", "keyboard", "pendrive","wire", "laptop", "Airconditioner"]
print(products)
```

```python
sample_product=("Mouse", 500, "Electronic")
type(sample_product)
```

```python
print(f"2nd and last product of list is: {products[1]} and {products[-1]}")
```

```python
products.append("Mobile")
products.append("Tablet")
print(f"After adding two products, the list is: {products}")
```

```python
converted_list= list(sample_product)
print(type(converted_list))
print(converted_list)
```

```python
converted_list[1]= 600
print(f"After changing the price of product, the list is: {converted_list}")
```

```python
updated_tuple= tuple(converted_list)
print(f"After converting the list back to tuple, the tuple is: {updated_tuple}")
```
