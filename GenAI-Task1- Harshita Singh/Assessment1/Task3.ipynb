```python
price_dict={"Laptop": 1000, "Mobile": 500, "Headphones": 100, "Keyboard": 50, "iphone": 100000, "ipad": 800.50, "Camera": 700, "Speaker": 150}
print(f"Original dictionary: {price_dict}")
print(len(price_dict))
```

```python
price_dict["Tablet"] = 300
print(f"After adding a new product, the updated dictionary is: {price_dict}")
new_length=len(price_dict)
print(f"Number of products after adding a new product: {new_length}")
```

```python
avg_price=sum(price_dict.values())/len(price_dict)
print(f"Average price of products: {round(avg_price, 2)}")
```

```python
max_price= max(price_dict.values())
min_price=min(price_dict.values())
print(f"Maximum price of products: {max_price}")
print(f"Minimum price of products: {min_price}")
```
