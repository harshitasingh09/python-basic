```python
products = [
    "laptop",
    "Mouse",
    "Banana",
    "Shampoo",
    "KeyError"
]

price_dict = {"laptop": 50000,"Mouse": 500,"Banana": 60,"Shampoo": 250,"KeyError": 1000}

category_dict = {"laptop": "Electronics","Mouse": "Electronics","Banana": "Fruits","Shampoo": "Personal Care","KeyError": "Electronics"}
```

```python
catalog = []

for item in products:
    catalog.append(((item, price_dict[item], category_dict[item])))

print(catalog)
```
