```python
orders = []

while True:
    print("\nMenu")
    print("1 - Add order amount")
    print("2 - Show all orders and totals after applying discounts")
    print("q - Quit")

    choice = input("Enter your choice: ")

    if choice == "1":
        amount = float(input("Enter order amount: "))
        orders.append(amount)

    elif choice == "2":
        if len(orders) == 0:
            print("No orders available.")
            continue

        total = 0

        print("\nOrders after discount:")
        for order in orders:
            if order >= 100:
                discounted = order * 0.90   
            elif order >= 50:
                discounted = order * 0.95  
            else:
                discounted = order         

            print(f"Original: ${order:.2f} -> After Discount: ${discounted:.2f}")
            total += discounted

        print(f"Total after discounts: ${total:.2f}")

    elif choice.lower() == "q":
        print("exiting!")
        break

    else:
        print("Invalid choice. Please try again.")
        continue
```
