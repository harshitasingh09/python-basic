```python
order_amount = input("Enter the order amount: ")
```

```python
if not order_amount.isdigit():
    print("Error: Please enter a valid number.")
else:
    order_amount = int(order_amount)
```

```python
if order_amount < 0:
    print("Error: Please enter a valid amount.")
  
else:
    discount=0
    if order_amount >= 2000:
        discount= order_amount * 15/100
        print(discount)
    elif order_amount >= 1500: 
        discount= order_amount * 10/100
        print(discount)
    elif order_amount >= 1000:
        discount= order_amount * 7/100
        print(discount)
    else:      
        print(discount)
   
total = order_amount - discount
subtotal = total + (total * 5/100)
print(subtotal)
```
