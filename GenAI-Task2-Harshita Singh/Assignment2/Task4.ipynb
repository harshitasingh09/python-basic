```python
daily= [200,150,0,400,50,-1,300]
total_sales=0
for sale in daily:
    if sale <0 :
        print("corrupt data found, exiting")
        break
    elif sale==0:
        print("no sales")
        continue
    else:
        total_sales+=sale
        print(f"running total sales is {total_sales}")
print(f"Total Sales: {total_sales}")
```
