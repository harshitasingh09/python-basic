```python
def add_price(prices_list, price):
    prices_list.append(price)
```

```python
def get_average_price(prices_list):
    total = 0
    
    for price in prices_list:
        total = total + price
    
    return total / len(prices_list)
```

```python
def get_max_price(prices_list):
    maximum = prices_list[0]
    
    for price in prices_list:
        if price > maximum:
            maximum = price
    
    return maximum
```

```python
prices = [100, 250, 500]

while True:
    print("\n--- Price Menu ---")
    print("1 → Add price")
    print("2 → Show average price")
    print("3 → Show highest price")
    print("q → Quit")

    choice = input("Enter your choice: ")

    if choice == "1":
        price = float(input("Enter price: "))
        add_price(prices, price)
        print("Price added successfully.")

    elif choice == "2":
        average = get_average_price(prices)
        print("Average price:", average)

    elif choice == "3":
        maximum = get_max_price(prices)
        print("Highest price:", maximum)

    elif choice == "q":
        print("Goodbye!")
        break

    else:
        print("Invalid choice.")


```
