```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}
```

```python
discount = float(input("Enter discount percentage: "))
with open("discount_report.txt", "w") as file:

    file.write("Product | Original Price | Discounted Price\n")
    file.write("-------------------------------------------\n")

    total_discounted_price = 0
    for product, original_price in prices.items():
        discounted_price = original_price - (original_price * discount / 100)

        file.write(
            product + " | " +
            str(original_price) + " | " +
            str(round(discounted_price, 2)) + "\n"
        )

        total_discounted_price += discounted_price
        average_discounted_price = total_discounted_price / len(prices)

    # Optional summary
    file.write("\n")
    file.write("Total Items: " + str(len(prices)) + "\n")
    file.write(
        "Average Discounted Price: " +
        str(round(average_discounted_price, 2))
    )

# Read the file and print it
print("\nDiscount Report:")
with open("discount_report.txt", "r") as file:
    content = file.read()

print(content)
```
