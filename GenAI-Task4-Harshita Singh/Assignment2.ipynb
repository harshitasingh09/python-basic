```python
with open("sales_data.txt", "r") as file:
    content = file.read()

print("Using read():")
print(content)
```

```python
with open("sales_data.txt", "r") as file:
    first_line = file.readline()

print("Using readline():")
print(first_line.strip())
```

```python
with open("sales_data.txt", "r") as file:
    lines = file.readlines()

sales = [int(line.strip()) for line in lines]

print("Using readlines():")
print(sales)
```
