```python
product1 = input("Enter first product name: ")
price1 = input("Enter first product price: ")

product2 = input("Enter second product name: ")
price2 = input("Enter second product price: ")

product3 = input("Enter third product name: ")
price3 = input("Enter third product price: ")
```

```python
with open("products.txt", "w") as file:
    file.write(product1 + " | " + price1 + "\n")
    file.write(product2 + " | " + price2 + "\n")
    file.write(product3 + " | " + price3 + "\n")
```

```python
print("\nProduct Information:")

with open("products.txt", "r") as file:
    for line in file:
        print(line.strip())
```
